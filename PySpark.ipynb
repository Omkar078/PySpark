### GroupBy Operation


```python
### Count Employees Department Wise
df.groupBy('department').count().show()
```

```python
### Average Salary Department Wise'
df.groupby('department').avg("salary").show()
```

```python
### Maximum Salart Department Wise
df.groupBy("department").max("salary").show()
```

```python
### Minimum Salary Department Wise
df.groupBy("department").min("salary").show()
```

```python
#### Total Salary Department Wise
df.groupBy("department").sum("salary").show()
```

```python
#### Multiple Aggregations
from pyspark.sql.functions import count, avg, max, min, sum

df.groupBy("department").agg(
    count("*").alias("Total Employees"),
    avg("salary").alias("Average Salary"),
    max("salary").alias("Maximum Salary"),
    min("salary").alias("Minimum Salary"),
    sum("salary").alias("Total Salary")
).show()
```
