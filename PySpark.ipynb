```python
#### Find the highest-paid employee under each manager.
from pyspark.sql.window import Window
from pyspark.sql.functions import col, desc, row_number

df.join(dept_df, on="department", how="inner") \
  .withColumn("rank", row_number().over(Window.partitionBy("manager_name").orderBy(desc("salary")))) \
  .filter(col("rank") == 1) \
  .drop("rank") \
  .select(dept_df["manager_name"], df["name"], df["salary"]) \
  .show()
```

```python
##Find the average salary of employees working on each project.
df.join(
    project_df,
    on = "emp_id",
    how = "inner"
) \
    .groupby(project_df["project_name"]).agg(
        round(avg(df["salary"]), 2).alias("average_salary")
    ) \
    .select(project_df["project_name"], col("average_salary")) \
    .show()
```

```python
###Find the manager who has the highest average employee performance score
df.join(
    dept_df,
    on = "department",
    how = "inner"
) \
    .join(
        performance_df,
        on = "emp_id",
        how = "inner"
    ) \
        .groupby(dept_df["manager_name"]) \
            .agg(
                round(avg(df["salary"]), 2) \
                    .alias("average_salary")
            ) \
                .orderBy(desc("average_salary")) \
                .limit(1) \
                .show()
```

```python
####  Find the number of employees in each state after joining the location table.
df.join(
    location_df,
    on = "city",
    how = "inner"
) \
    .groupby(location_df["state"]) \
        .agg(
            count(df["emp_id"]) \
                .alias("emp_cnt")
        ) \
            .show()
```

```python
#### Find employees who are Active, have a rating of A, and work in the Technology division.
df.join(
    performance_df,
    on = "emp_id",
    how = "inner"
).join(
    dept_df,
    on = "department",
    how = "inner"
).filter(
    (df["status"] == "Active") & 
    (performance_df["rating"] == "A") & 
    (dept_df["division"] == "Technology")
).show()
```

### Join + Filter + Aggregations

```python
#### Find all Active employees who work in the Technology division.

df.join(
    dept_df,
    on = "department",
    how = "inner"
) \
    .filter(
        (df["status"] == "Active")
        &(dept_df["division"] == "Technology")
    ) \
    .select(
        "emp_id",
        "name",
        "department",
        "division",
        "salary",
        "status"
    ) \
    .show()

```

```python
###Find employees who: Are Active, Salary > 90,000 Work in Technology
df.join(
    dept_df,
    on = "department",
    how = "inner"
) \
    .filter(
        (df["status"] == "Active")
        &(dept_df["division"] == "Technology")
        &(df["salary"] > 90000)
    ) \
    .select(
        "name",
        "department",
        "salary",
        "manager_name"
    ) \
    .show()
```

```python
#### Find the number of employees under each manager.
df.join(
    dept_df,
    on = "department",
    how = "inner"
) \
.groupby(dept_df["manager_name"]) \
.agg(
    count(df["emp_id"]) \
    .alias("employee_count")
) \
.show()
```

```python
##Find the following for each manager:
# Employee count
# Average salary
# Maximum salary
# Minimum salary

df.join(
    dept_df,
    on = "department",
    how = "inner"
) \
.groupby(dept_df["manager_name"]) \
.agg(
    count(df["emp_id"]) \
    .alias("employee_count"),
    round(avg(df["emp_id"]),2) \
    .alias("avg_salary"),
    max(df["emp_id"]) \
    .alias("max_salary"),
    min(df["emp_id"]) \
    .alias("min_salary")
) \
.show()
```

```python
#### Find the average salary of Active employees for each department.
df.join(
    dept_df,
    on =  "department",
    how = "inner"
) \
.groupby(dept_df["department"]) \
.agg(
    round(avg(df["salary"]), 2).alias("avg_salary"))  \
.show() 
```
